### <center>  VWAP — Volume Weighted Average Price </center>

---

#### 1. Définition et intuition du VWAP

Le VWAP (Volume Weighted Average Price) représente le prix moyen d’un instrument pondéré par les volumes échangés sur une période donnée. 
Il mesure le prix auquel la majorité des transactions du marché ont eu lieu.

Le VWAP est à la fois :
- une mesure de référence pour évaluer la qualité d’exécution,
- une base pour construire des algorithmes d’exécution passifs.

En exécution algorithmique, un algo VWAP répartit un ordre important sur la journée en suivant la distribution moyenne du volume du marché (plus actif aux heures où le volume est élevé), de façon à obtenir un prix d’exécution moyen proche (ou meilleur) que le VWAP du marché.

- Pour un **acheteur** → objectif : payer **≤ VWAP_market**
- Pour un **vendeur** → objectif : vendre **≥ VWAP_market**

Caractéristiques de l’algorithme
> - **Benchmark-based** : la performance est évaluée par rapport au VWAP du marché(benchmark)
> - **Schedule-based** : suit un plan (schedule) basé sur un **profil de volume intraday**
> - **Single-order** : travaille sur **un seul ordre parent** (un instrument à la fois)



#### 2. Calcul de l'indicateur VWAP

Le VWAP  se définit par : $VWAP = \frac{\sum_i P_i \times V_i}{\sum_i V_i}$ où :

- $P_i$ : prix de la transaction $i$
- $V_i$ : volume échangé lors de la transaction $i$

#### Exemple

Imaginons qu’un gérant d’un fonds souhaite acheter 200 000 actions TotalEnergies dans la journée, entre 9h et 17h.
S’il envoie tout l’ordre d’un coup, il risque de :
- Faire monter le prix (impact de marché)
- Et donc payer plus cher que le marché
  
Pour éviter cela, il confie l’ordre à un algorithme VWAP, dont l’objectif est d’obtenir un prix d’exécution moyen proche du VWAP du marché. L’idée est de répartir l’ordre total sur la journée selon la liquidité naturelle du marché.

**Paramètres de l’ordre parent (Ordre global confié à l'algo)** :
- Quantité totale : Q = 200 000
- Sens : achat
- Période d’exécution : 9h–17h
- Benchmark : VWAP_market

**Construction du Schedule (plan d’exécution)**
- Le VWAP s’appuie sur un **profil de volume intraday** du titre.
- Exemple : 30 % du volume le matin, 50 % à midi, 20 % en fin de journée 
- L’algorithme répartit alors la quantité totale selon ce profil :

| Période | % Volume | Quantité à exécuter |
|----------|-----------|--------------------|
| 9h–11h | 30 % | 60 000 |
| 11h–15h | 50 % | 100 000 |
| 15h–17h | 20 % | 40 000 |

**Envoi d’ordres enfants (Child Orders)**

L’algorithme envoie progressivement des ordres enfants (`child orders`) :
- Chaque tranche `q_t` est exécutée à son horaire prévu
- Souvent via **ordres limites** ou **passifs**
- Le rythme peut s’ajuster selon le **volume réel du marché**

Les child orders sont ensuite soumis au moteur d’exécution avec des contraintes comme :
- Taille maximale par ordre
- Distance de prix par rapport au mid-price
- Fréquence d’envoi / refresh
